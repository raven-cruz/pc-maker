```python
import requests
import json
from google.oauth2 import service_account
from google.auth.transport.requests import Request
from IPython.display import Audio
import base64

keys_loc = "../keys"
```

```python
# Load credentials from your service account file
creds = service_account.Credentials.from_service_account_file(
    f"{keys_loc}/service.json",
    scopes=["https://www.googleapis.com/auth/cloud-platform"],
)
print("TEST")
```

```python
# Fetch an access token (this is what gcloud auth print-access-token returns)
creds.refresh(Request())
token = creds.token
```

```python
with open(f'{keys_loc}/keys.json', 'r') as f:
    keys = json.load(f)
```

```python
def get_audio(subject):

    url = "https://aiplatform.googleapis.com/v1beta1/projects/wpp-cto-smartprod-sandbox/locations/global/interactions"

    payload = {
      "model": "lyria-3-clip-preview",
      "input": [
        {
          "type": "text",
          "text": f"everyone is so fried and done with {subject}"
        }
      ]
    }
    headers = {
        "Authorization": f"Bearer {token}",
        "Content-Type": "application/json",
    }

    response = requests.post(url, headers=headers, json=payload)
    data = response.json()
    return data["outputs"][2]["data"]
```

```python
my_song = get_audio("class")
my_song
```

```python
mp3_bytes = base64.b64decode(my_song)
Audio(mp3_bytes)
```
